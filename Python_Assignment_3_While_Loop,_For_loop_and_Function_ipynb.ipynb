Problem Statement: Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range

A for loop is used to iterate over a sequence such as a list, string, tuple, or range.

```python
# 1. Prompt user for Input. Ask the user to enter a number for which they want to generate a multiplication table.

number = int(input("enter a number  to generate a multiplication table : "))
```

```python
# 2. Generate the Multiplication Table: Use a for loop to iterate through the numbers 1 to 10. In each iteration, calculate the product of the user's number and the current number from the loop.

for i in range (1,11):
   print (f"{number} * {i} = {number * i}")

```

```python
# 3. Display the Multiplication Table:
# Print each line of the multiplication table in the format: "number x i = result".

number = int(input("Enter the number to generate a multiplication table: "))

for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")
```

# **Function:**
### ***BMI Calculator***

Problem Statement: Create a Python program that calculates the Body Mass Index
(BMI).

Hint: BMI = weight (kg) / [height (m)]²

```python
# Instructions:
# 1. Define a function calculate_bmi(weight, height) that returns the BMI.

def calculate_bmi(weight, height):  # def defines the function.
    bmi = weight / (height ** 2)
    return bmi  #  gives the value back to the caller so it can be stored or used later


# 2. Prompt the user for their weight (in kg) and height (in meters).

weight = float(input("Enter your weight (in kg): ")) # float converts the text into a number with decimals.
height = float(input("Enter your height (in meters): "))


# 3. Use the function to calculate and display the BMI.

bmi = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi:f}") # displays the value on the screen.

```

```python
# output

def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

weight = float(input("Enter your weight (in kg): "))
height = float(input("Enter your height (in meters): "))

bmi = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi:f}")
```
